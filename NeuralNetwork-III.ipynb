```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
class CustomNeuralNetwork(keras.Model):
    def __init__(self, num_classes=10, **kwargs):
        super(CustomNeuralNetwork,self).__init__(**kwargs)
        self.dense1 = layers.Dense(128,activation='relu',name='hidden_layer_1')
        self.dense2 = layers.Dense(128,activation='relu',name='hidden_layer_2')
        self.dropout = layers.Dropout(0.2)
        self.classifier = layers.Dense(num_classes,activation='softmax',name='output_layer')

    def call(self,inputs,training=False):
        x = self.dense1(inputs)
        x = self.dense2(x)

        if training:
            x = self.dropout(x,training=training)
        return self.classifier(x)            
```

```python
subclassed_model = CustomNeuralNetwork(num_classes=10,name="my_subclassed_model")
subclassed_model.build(input_shape=(None,784))
subclassed_model.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```
