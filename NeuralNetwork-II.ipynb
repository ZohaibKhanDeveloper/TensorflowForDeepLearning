Neural Network Using Functional API

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
inputs = keras.Input(shape=(784,),name='input_layer')
x = layers.Dense(128,activation='relu',name='hidden_layer')(inputs)
x = layers.Dense(64,activation='relu',name='hidden_layer_2')(x)
outputs = layers.Dense(10,activation='softmax',name='output_layer')(x)
```

```python
functional_model = keras.Model(inputs=inputs,outputs=outputs,name='my_functional_model')
functional_model.compile(optimizer='adam',loss='sparse_categorical_cross_entropy',metrics=['accuracy'])
functional_model.summary()
```
