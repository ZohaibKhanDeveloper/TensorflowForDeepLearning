```python
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential([
    Dense(3,input_shape=(3,),activation='relu'),
    Dense(1)
])
model.summary()
model.compile(optimizer='adam',loss=tf.keras.losses.MAE,metrics=['mae'])
df = pd.read_csv("./custom_datasets/housing_data.csv")
x_train = df[['age', 'income', 'rooms']].to_numpy()
y_train = df['house_price'].to_numpy()
dataset = tf.data.Dataset.from_tensor_slices((x_train, y_train))
dataset = dataset.shuffle(buffer_size=1024).batch(32)
model.fit(dataset, epochs=10)
```

```python
import numpy as np
model.predict(np.array(([[1,2,3]])))
```
