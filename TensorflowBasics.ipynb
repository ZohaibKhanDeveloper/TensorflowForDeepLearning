<h2>Tensorflow Simple Functions</h2>

```python
import tensorflow as tf
```

```python
x = tf.constant(4,shape=(1,1),dtype=tf.float64)
print(x)
```

```python
x = tf.constant([[1,2,3],[4,5,6]])
print(x)
```

```python
x = tf.ones((3,3))
print(x)
```

```python
x = tf.zeros((2,3))
print(x)
```

```python
x = tf.eye(3)
print(x)
x = tf.eye(2)
print(x)
```

```python
x = tf.random.normal((3,3),mean=0,stddev=1)
print(x)
```

```python
x = tf.random.uniform((1,3),minval=0,maxval=1)
print(x)
```

```python
x = tf.range(9)
print(x)
```

```python
x = tf.range(start=1,limit=10,delta=2)
print(x)
```

```python
x = tf.cast(x,dtype=tf.float64)
print(x)
```

<h2>Mathematical Operations</h2>

```python
x = tf.constant([1,2,3])
y = tf.constant([4,5,6])
sum = tf.add(x,y)
print(x,"\n",y,"\n",sum)
```

```python
difference = tf.subtract(y,x)
print(difference)
```

```python
difference = y-x
print(difference)
```

```python
x = tf.constant([3,8,10])
y = tf.constant([6,16,20])
```

```python
division = y / x
print(division)
division = tf.divide(y,x)
print(division)
```

```python
multiplication = tf.multiply(x,y)
print(multiplication)
multiplication = x*y
print(multiplication)
```

```python
z = tf.tensordot(x,y,axes=1)
print(z)
```

```python
z = tf.reduce_sum(x*y,axis=0)
print(z)
```

```python
a = tf.constant([[1,2,3],[4,5,6],[7,8,9]])
b = tf.constant([[1,2,3],[4,5,6],[7,8,9]])
matrix_product = tf.matmul(a,b)
print(matrix_product)
```

```python
print(a,"\n",b)
```

```python
a = tf.constant([[1,2,3],[4,5,6]])
print(a)
```

```python
a = tf.reshape(a,(3,2))
print(a)
```

```python
a = tf.transpose(a)
print(a)
```
